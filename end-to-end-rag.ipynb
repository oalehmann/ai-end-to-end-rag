```python
## Verificación de que los txt fueron creados correctamente
files = list(DATA_DIR.glob("*.txt"))

print("Documentos encontrados:", len(files))

for file in files:
    content = file.read_text(encoding="utf-8")
    print(f"{file.name}: {len(content)} caracteres")
```

```python
## Contamos los tokens de los documentos
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

for file in files:
    content = file.read_text(encoding="utf-8")
    tokens = len(encoding.encode(content))

    print(f"{file.name}: {tokens} tokens")
```

```python
## Preparación de files txt para que python los trabaje (los transformamos en diccionarios)

documents = []

for file in files:

    content = file.read_text(encoding="utf-8")

    documents.append({
        "source": file.name,
        "content": content
    })

    print("Documentos cargados:", len(documents))

print(type(documents))
print(type(documents[0]))
print(documents[0].keys())
```

```python
## Vamos a pasar la información del diccionario al formato Documents que es propio de langchain y me da info de metadat

from langchain_core.documents import Document

langchain_documents = []

for document in documents:

    langchain_document = Document(
        page_content=document["content"],
        metadata={
            "source": document["source"]
        }
    )

    langchain_documents.append(langchain_document)

print(type(langchain_documents))
print(len(langchain_documents))

print(type(langchain_documents[0]))

print(langchain_documents[0].page_content[:200])
print(langchain_documents[0].metadata)
```

```python
## Creamos el objeto splitter y luego lo aplicamos al langchain_documents y generamos los chunks

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base",
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(langchain_documents)
```

```python
## Verificaciones de la generación de chunks
print("Total de chunks:", len(chunks))

## Verificamos primer chunk
print(type(chunks[0]))

print("\nContenido:")
print(chunks[0].page_content)

print("\nMetadata:")
print(chunks[0].metadata)
```

```python
## Creamos instancia para armar embedings

from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)
```

```python
## Armamos instancia de Chroma

from langchain_chroma import Chroma

VECTORSTORE_DIR = PROJECT_DIR / "vectorstore"

vectorstore = Chroma(
    collection_name="rag_knowledge_base",
    embedding_function=embeddings,
    persist_directory=str(VECTORSTORE_DIR)
)
```

```python
## Generamos la base de datos vectorial

vectorstore.add_documents(chunks)

vectorstore._collection.count()
```

```python
## Armado de función de encapuslación de la pregunta al LLM

async def get_rag_response(query: str):

      results = vectorstore.similarity_search(
        query,
        k=3
    )
      context = "\n\n".join(
        result.page_content
        for result in results
    )


      references = list(dict.fromkeys(
        result.metadata["source"]
        for result in results
))


      question = query

      prompt = f"""
      Respondé la pregunta utilizando exclusivamente el contexto proporcionado.

      Si el contexto no contiene información suficiente, respondé:
      "No encuentro información suficiente en el contexto para responder."

      Contexto:
     {context}

      Pregunta:
      {question}
     """

      response = await llm.ainvoke(prompt)

      return RAGResponse(
        respuesta=response.content,
        referencias=references
    )


```

```python
## prueba básica de la función

answer = await get_rag_response(
    "¿Qué permite la migración en vivo?"
)

print(answer)
```

```python
## Creación de función de ingesta ingestion.py

ingestion_code = '''
from pathlib import Path

from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma


def create_vectorstore(
    data_dir: str,
    vectorstore_dir: str
):

    data_path = Path(data_dir)
    vector_path = Path(vectorstore_dir)

    documents = []

    for file_path in data_path.glob("*"):
        if file_path.suffix in [".txt", ".md"]:

            content = file_path.read_text(
                encoding="utf-8"
            )

            documents.append(
                Document(
                    page_content=content,
                    metadata={"source": file_path.name}
                )
            )

    text_splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
        encoding_name="cl100k_base",
        chunk_size=500,
        chunk_overlap=50
    )

    chunks = text_splitter.split_documents(documents)

    embeddings = OpenAIEmbeddings(
        model="text-embedding-3-small"
    )

    vectorstore = Chroma(
        collection_name="rag_knowledge_base",
        embedding_function=embeddings,
        persist_directory=str(vector_path)
    )

    if vectorstore._collection.count() == 0:
        vectorstore.add_documents(chunks)

    return vectorstore
'''

ingestion_path = PROJECT_DIR / "ingestion.py"

ingestion_path.write_text(
    ingestion_code.strip() + "\n",
    encoding="utf-8"
)

print("Archivo creado:", ingestion_path)
```

```python
## Armamos file rag_chain.py

rag_chain_code = '''
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_chroma import Chroma
from schemas import RAGResponse


def create_rag_chain(vectorstore_dir: str):

    embeddings = OpenAIEmbeddings(
        model="text-embedding-3-small"
    )

    vectorstore = Chroma(
        collection_name="rag_knowledge_base",
        embedding_function=embeddings,
        persist_directory=vectorstore_dir
    )

    llm = ChatOpenAI(
        model="gpt-4o-mini",
        temperature=0
    )

    async def get_rag_response(query: str) -> RAGResponse:

        results = vectorstore.similarity_search(
            query,
            k=3
        )

        context = "\\n\\n".join(
            result.page_content
            for result in results
        )

        references = list(dict.fromkeys(
            result.metadata["source"]
            for result in results
        ))

        prompt = f"""
        Respondé la pregunta utilizando exclusivamente
        la información del siguiente contexto.

        Si el contexto no contiene información suficiente,
        respondé exactamente:
        No encuentro información suficiente en el contexto para responder.

        CONTEXTO:
        {context}

        PREGUNTA:
        {query}
        """

        response = await llm.ainvoke(prompt)

        return RAGResponse(
            respuesta=response.content,
            referencias=references
        )

    return get_rag_response
'''

rag_chain_path = PROJECT_DIR / "rag_chain.py"

rag_chain_path.write_text(
    rag_chain_code.strip() + "\n",
    encoding="utf-8"
)

print("Archivo creado:", rag_chain_path)
```

```python
## Armamos main.py

main_code = '''
import asyncio
from pathlib import Path

from rag_chain import create_rag_chain


PROJECT_DIR = Path(__file__).parent
VECTORSTORE_DIR = PROJECT_DIR / "vectorstore"


async def main():

    get_rag_response = create_rag_chain(
        vectorstore_dir=str(VECTORSTORE_DIR)
    )

    query = input("Ingresá tu pregunta: ")

    result = await get_rag_response(query)

    print("\\nRESPUESTA:")
    print(result.respuesta)

    print("\\nREFERENCIAS:")
    for reference in result.referencias:
        print(f"- {reference}")


if __name__ == "__main__":
    asyncio.run(main())
'''

main_path = PROJECT_DIR / "main.py"

main_path.write_text(
    main_code.strip() + "\n",
    encoding="utf-8"
)

print("Archivo creado:", main_path)
```
